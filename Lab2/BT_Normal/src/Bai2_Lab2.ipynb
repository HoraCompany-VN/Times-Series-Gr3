# Bài 2: Kiểm định ADF trên chuỗi gốc


## 1. Load dữ liệu

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller

df = pd.read_csv('a10.csv', parse_dates=['date'], index_col='date')
print(df.head())
print('Số quan sát:', len(df))

df['value'].plot(figsize=(10, 4), title='Chuỗi gốc a10')
plt.ylabel('value')
plt.show()
```

## 2. Kiểm định ADF trên chuỗi gốc

```python
result = adfuller(df['value'], autolag='AIC')

adf_stat, p_value, used_lag, nobs, crit_values = result[0], result[1], result[2], result[3], result[4]

print(f'ADF statistic : {adf_stat:.4f}')
print(f'p-value       : {p_value:.4f}')
print(f'Số độ trễ     : {used_lag}')
print(f'Số quan sát   : {nobs}')
print('Giá trị tới hạn:')
for level, value in crit_values.items():
    print(f'   {level}: {value:.4f}')

# Kết luận tự động
if p_value < 0.05:
    print('\nKết luận: Bác bỏ H0 -> Chuỗi DỪNG')
else:
    print('\nKết luận: Không bác bỏ H0 -> Chuỗi KHÔNG DỪNG')
```

## 4. Trả lời câu hỏi

### Câu 1. H₀ của ADF là gì?

- **H₀:** Chuỗi có nghiệm đơn vị (unit root), tức là **chuỗi không dừng**.
- **H₁:** Chuỗi **dừng**.

ADF giả định ngay từ đầu là chuỗi *không dừng*. Muốn kết luận chuỗi dừng thì phải có đủ bằng chứng để bác bỏ H₀.

### Câu 2. p-value cho biết điều gì?

p-value là xác suất thu được kết quả như trên **nếu H₀ đúng** (chuỗi thật sự không dừng).

- p < 0,05 → bác bỏ H₀ → chuỗi **dừng**
- p ≥ 0,05 → không bác bỏ H₀ → chuỗi **không dừng**

Ở đây p = 1,0, lớn hơn rất nhiều so với 0,05, nên hoàn toàn không có bằng chứng chuỗi dừng.

### Câu 3. Chuỗi có dừng không?

**Không.** Hai cách kiểm tra cho cùng kết quả:

1. p-value = 1,0 > 0,05.
2. ADF statistic = 3,145 **lớn hơn** mọi giá trị tới hạn. Muốn bác bỏ H₀ thì ADF phải *âm hơn* giá trị tới hạn (ví dụ < −2,88 ở mức 5%).

Điều này khớp với đồ thị: chuỗi có **xu hướng tăng rõ** (từ khoảng 3 năm 1991 lên khoảng 20–30 năm 2008) và **mùa vụ** (tháng 1 cao vọt, tháng 2 tụt mạnh). Trung bình thay đổi theo thời gian nên chuỗi không dừng → Bài 3 cần lấy sai phân.

> **Mẹo nhớ:** ADF càng âm thì chuỗi càng có khả năng dừng. ADF dương như ở đây là dấu hiệu rất rõ của chuỗi không dừng.

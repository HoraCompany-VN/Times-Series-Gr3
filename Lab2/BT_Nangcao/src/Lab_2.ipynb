```python
import pandas as pd
from statsmodels.tsa.stattools import adfuller

df = pd.read_csv(r"C:\Users\thang\Downloads\airline-passengers.csv", parse_dates=["Month"], index_col="Month")
y = df["Passengers"]

result = adfuller(y)
print(f"ADF Statistic : {result[0]:.4f}")
print(f"p-value       : {result[1]:.4f}")
print(f"Lags used     : {result[2]}")
for k, v in result[4].items():
    print(f"Critical {k}  : {v:.4f}")

print("=> Dừng" if result[1] < 0.05 else "=> KHÔNG dừng")
```

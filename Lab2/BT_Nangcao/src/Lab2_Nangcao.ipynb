# Bài 1: Vẽ chuỗi thời gian và nhận diện mùa vụ
- Dữ liệu: Airline Passengers (1949 - 1960)
- Nguồn dữ liệu: https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv

```python
# Import thư viện và tải dữ liệu
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"
df = pd.read_csv(url)

df["Month"] = pd.to_datetime(df["Month"])
df.set_index("Month", inplace=True)

print("Kích thước dữ liệu:", df.shape)
print(df.head())
```

```python
# Vẽ biểu đồ chuỗi thời gian
plt.figure(figsize=(12, 5))
plt.plot(df.index, df["Passengers"], marker="o", markersize=3)
plt.title("Số lượng hành khách hàng không quốc tế (1949–1960)")
plt.xlabel("Thời gian")
plt.ylabel("Số lượng hành khách (nghìn người)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Nhận diện mùa vụ theo tháng
monthly_avg = df.groupby(df.index.month)["Passengers"].mean()

plt.figure(figsize=(10, 5))
plt.plot(monthly_avg.index, monthly_avg.values, marker="o")
plt.xticks(range(1, 13))
plt.title("Số lượng hành khách trung bình theo tháng")
plt.xlabel("Tháng")
plt.ylabel("Số hành khách trung bình (nghìn người)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### Nhận xét và Kết luận:
- **Xu hướng (Trend):** Dữ liệu có xu hướng tăng rõ rệt và liên tục từ năm 1949 đến 1960.
- **Tính mùa vụ (Seasonality):** Chuỗi thời gian có sự lặp lại biến động theo chu kỳ hàng năm. Cụ thể, mỗi năm đều có các đỉnh cao (thường vào các tháng hè) và đáy thấp (thường vào cuối/đầu năm).
- **Biên độ dao động (Variance):** Biên độ của các dao động mùa vụ tăng dần theo thời gian (khoảng cách giữa đỉnh và đáy ngày càng rộng ra), cho thấy đây có thể là chuỗi thời gian dạng nhân (Multiplicative).
- **Tính dừng:** Chuỗi ban đầu **không dừng** (non-stationary) vì trung bình (mean) và phương sai (variance) thay đổi theo thời gian.
